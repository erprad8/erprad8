```python
#
'''
indexing --- used to or process of accessing a single element from a list

1.positive (starts from 0 at beg)
2.negative (start from -1 from last )

'''
```

```python
l1=['g','hello',9,10,'orange']
# print(l1)
print(l1[1]) #positive indexing
print(l1[-4]) #negative indexing
print(l1[4])
print(l1[-1])
```

```python
'''
slicing --  it is aprocess of accessing multiple eleemst based on index

syntax --- list_name[start:stop(n-1):step]

start -- stating index
stop -- ending index(n-1) -- n is number of elements
step -- step size(gap) ---- optional parameter  default step is 1

'''
```

```python
l2=['g','hello',9,10,'orange']
# print(l2)
# print(l2[1:3])
# print(l2[-4:-1])
# print(l2[0:])
# print(l2[-1])
print(l2[0:5:3]) #(start:stop:step)
print(l2[:-1])
print(l2[1:6:2])
print(l2[-1:])
```

```python
#reverse this list
print(l2[::-1])

'''
start -- start from starting
stop --- go till end
step --- do backtrack

'''
```

```python
#append - to add element at end of list
numbers=[10,20,30]
print(numbers)
numbers.append([40,50]) #it will treat as single element
print(numbers)
```

```python
#extend - adds multiple elemenst to end
numbers=[10,20,30]
print(numbers)
numbers.extend([40,50,60])
print(numbers)
```

```python
#insert -- adds an elements at aspecific index
numbers=[10,20,30]
print(numbers)
numbers.insert(1,26)
numbers.insert(2,30) #using index
print(numbers)
```

```python
#remove -- remove the specific value
numbers=[10,20,30]
print(numbers)
numbers.remove(20)
print(numbers)
```

```python
#pop -- remove and return an element
numbers=[10,20,30]
print(numbers)
x=numbers.pop()
print(x)
print(numbers)
```

```python
#clear -- removes all elemenst in list
l1=[10,20,30,40,50,60]
print(l1)
l1.clear()
print(l1)
```

```python
#index -- return the index position of the element
l1=[10,20,30,40,50,60]
print(l1)
print(l1.index(30))
```

```python
#count() -- this will count the how many time the value appears
numbers=[10,20,30,10,10,20,30,40,50]
print(numbers.count(30))
```

```python
#sort() -- used to arrange the element in asc and desc order (DEFAULT ASC)

numbers=[30,10,5,3,60]
numbers.sort(reverse=True) #for desc order
print(numbers)
```

```python
numbers=[30,10,5,3,60]
numbers.sort()
print(numbers)
```

```python
#reverse()--reverse the existing list

numbers=[10,20,30,40,50]
print(numbers)
numbers.reverse()
print(numbers)
```

```python
#copy -- it copy the element of existong list to new list
numbers=[10,20,30,40,50]
new_numbers=numbers.copy()
print(new_numbers)
```

```python
#len -- returns the number of elemenst

numbers=[10.1,20.2,30.3,40,50,60,70]
print(len(numbers))
print(max(numbers))
print(min(numbers))
print(sum(numbers))
```
